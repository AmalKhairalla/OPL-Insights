# DS330 Project Part 1 — Train and Tune a Deep Neural Network

## Predicting Powerlifting Performance

**Dataset:** OpenPowerlifting — FPR Single-ply

**Task:** Predicting best deadlift performance (kg) using athlete and competition features.

**Objective:** Train and compare a linear regression baseline with deep neural networks, and investigate how training configurations influence model performance and generalization.

This notebook contains the official training and tuning experiments. Dataset exploration and selection were documented separately.

```python
import pandas as pd
import numpy as np

url = (
    "https://raw.githubusercontent.com/"
    "AmalKhairalla/OPL-Insights/main/"
    "data/FPR_SinglePly_SBD.csv.gz"
)

df = pd.read_csv(url, compression="gzip")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python

# Display all column names
print("Available columns:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

# Display data types and missing values
feature_summary = pd.DataFrame({
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isna().sum(),
    "Missing (%)": (
        df.isna().mean() * 100
    ).round(2),
    "Unique Values": df.nunique()
})

display(feature_summary)
```

```python
print(f"\n{"Sex"}:")
display(df["Sex"].value_counts(dropna=False))
```

```python

import numpy as np
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit

# Keep a stable row index for saving split assignments
df = df.reset_index(drop=True).copy()

# Group by recorded athlete name
assert df["Name"].notna().all()

groups = df["Name"].astype(str).str.strip()
assert groups.ne("").all()

# Step 1: Reserve approximately 15% for testing
split_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=42
)

trainval_idx, test_idx = next(
    split_test.split(df, groups=groups)
)

# Step 2: Split the remaining groups into
# approximately 70% training and 15% validation
split_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15 / 0.85,
    random_state=43
)

train_rel, val_rel = next(
    split_val.split(
        df.iloc[trainval_idx],
        groups=groups.iloc[trainval_idx]
    )
)

train_idx = trainval_idx[train_rel]
val_idx = trainval_idx[val_rel]

train_df = df.iloc[train_idx].copy()
val_df = df.iloc[val_idx].copy()
test_df = df.iloc[test_idx].copy()

# Check whether names overlap between sets
train_names = set(train_df["Name"])
val_names = set(val_df["Name"])
test_names = set(test_df["Name"])

assert train_names.isdisjoint(val_names)
assert train_names.isdisjoint(test_names)
assert val_names.isdisjoint(test_names)

print("No recorded athlete names overlap!")

summary = pd.DataFrame({
    "Split": ["Training", "Validation", "Testing"],
    "Records": [
        len(train_df), len(val_df), len(test_df)
    ],
    "Unique names": [
        train_df["Name"].nunique(),
        val_df["Name"].nunique(),
        test_df["Name"].nunique()
    ]
})

summary["Records (%)"] = (
    summary["Records"] / len(df) * 100
).round(2)

display(summary)
```

```python

df["Split"] = ""

df.loc[train_idx, "Split"] = "train"
df.loc[val_idx, "Split"] = "validation"
df.loc[test_idx, "Split"] = "test"

df.to_csv(
    "/content/FPR_SinglePly_with_splits.csv.gz",
    index=False,
    compression="gzip"
)

print("Split assignments saved.")
```

```python

import pandas as pd

# Create a year column for exploration
df["Year"] = pd.to_datetime(
    df["Date"], errors="coerce"
).dt.year

# Count unique competition years per recorded name
athlete_history = (
    df.groupby("Name")
    .agg(
        Records=("Name", "size"),
        Different_Years=("Year", "nunique"),
        First_Year=("Year", "min"),
        Last_Year=("Year", "max")
    )
    .sort_values(
        ["Different_Years", "Records"],
        ascending=False
    )
)

display(athlete_history.head(10))
```

```python

# Select the recorded name with the most years
selected_name = athlete_history.index[0]

print("Selected athlete name:", selected_name)

# Display their competition history
athlete_records = df[
    df["Name"] == selected_name
].sort_values("Date")

columns_to_show = [
    "Name",
    "Year",
    "Age",
    "BodyweightKg",
    "Best3SquatKg",
    "Best3BenchKg",
    "Best3DeadliftKg"
]

display(
    athlete_records[columns_to_show]
)
```
