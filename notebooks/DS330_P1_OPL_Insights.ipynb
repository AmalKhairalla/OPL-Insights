## Decisions (fixed before coding)

| Decision | Value | Reason |
|---|---|---|
| Target | TotalKg | Continuous, many distinct values |
| Event filter | SBD only | Total means the same thing in every row |
| Features | Age, BodyweightKg, Year, Age_missing, Tested, Sex, Equipment | No leakage |
| Excluded | Federation, Country | Too many categories; project is about training |
| Split | Grouped by Name, 70/15/15, seed 42 | Same lifter has many similar rows; random split leaks |
| Sample | 200,000 rows, seed 42 (before split) | Course requirement |
| Epochs / seed | 30 / 42 in every run | Fair comparison |
| Target scaling | Standardized for training; MSE reported in kg² | Stable training, comparable numbers |

Leakage columns removed: Best3*Kg, Squat1-4Kg, Bench1-4Kg, Deadlift1-4Kg, Dots, Wilks, Glossbrenner, Goodlift, Place.

#The imports, seed, and device

```python
import os
import glob
import zipfile
import time
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GroupShuffleSplit

SEED = 42

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
```

#Download

```python
USE_COLS = ['Name', 'Sex', 'Event', 'Equipment', 'Age',
            'BodyweightKg', 'TotalKg', 'Tested', 'Date']

DTYPES = {
    'Sex': 'category',
    'Event': 'category',
    'Equipment': 'category',
    'Tested': 'category',
    'Age': 'float32',
    'BodyweightKg': 'float32',
    'TotalKg': 'float32',
}

df = pd.read_csv(CSV_PATH, usecols=USE_COLS, dtype=DTYPES)

print('Shape:', df.shape)
print(f'Memory: {df.memory_usage(deep=True).sum() / 1e6:.0f} MB')
display(df.head())
print(df.dtypes)
print(df.isna().mean().round(3))
```

#SBD filtration + cleaning log

```python
clean_log = []

def log_step(description, before, after):
    clean_log.append({'Step': description, 'Rows before': before,
                      'Rows after': after, 'Removed': before - after})
    print(f'{description}: {before:,} -> {after:,}  (removed {before - after:,})')

n0 = len(df)
df = df[df['Event'] == 'SBD'].copy()
log_step("Keep Event == 'SBD'", n0, len(df))

df = df.drop(columns='Event')
print(df.shape)
```

#The Target

```python
n0 = len(df)
df = df.dropna(subset=['TotalKg'])
log_step('Drop missing TotalKg', n0, len(df))

n0 = len(df)
df = df[df['TotalKg'] > 0]
log_step('Drop TotalKg <= 0', n0, len(df))
```

#Weight, gender, equipment

```python
n0 = len(df)
df = df.dropna(subset=['BodyweightKg'])
log_step('Drop missing BodyweightKg', n0, len(df))

print(df['Sex'].value_counts())
print()
print(df['Equipment'].value_counts())
```

#Tested and Date

```python
df['Tested'] = (df['Tested'] == 'Yes').astype('int8')

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Year'] = df['Date'].dt.year

n0 = len(df)
df = df.dropna(subset=['Year'])
log_step('Drop rows with unparsable Date', n0, len(df))

df['Year'] = df['Year'].astype('int16')
df = df.drop(columns='Date')

print(df['Tested'].value_counts())
print(df['Year'].describe())
```

#Outliers

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, ['Age', 'BodyweightKg', 'TotalKg']):
    ax.hist(df[col].dropna(), bins=60)
    ax.set_title(col)
plt.tight_layout()
plt.show()

AGE_MIN, AGE_MAX = 14, 90
BW_MIN, BW_MAX = 30, 250
TOTAL_MIN, TOTAL_MAX = 30, 1500

n0 = len(df)
df = df[df['Age'].isna() | df['Age'].between(AGE_MIN, AGE_MAX)]
log_step(f'Keep Age in [{AGE_MIN}, {AGE_MAX}] (missing kept)', n0, len(df))

n0 = len(df)
df = df[df['BodyweightKg'].between(BW_MIN, BW_MAX)]
log_step(f'Keep BodyweightKg in [{BW_MIN}, {BW_MAX}]', n0, len(df))

n0 = len(df)
df = df[df['TotalKg'].between(TOTAL_MIN, TOTAL_MAX)]
log_step(f'Keep TotalKg in [{TOTAL_MIN}, {TOTAL_MAX}]', n0, len(df))
```
