#Import Required Libraries

```python
import os
import zipfile

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

from google.colab import drive
drive.mount('/content/drive')
```

#Mount Google Drive & Define Dataset Path

```python
DATA_PATH = "/content/drive/MyDrive/DS330_Project/openpowerlifting-latest.zip"

print("Dataset path:")
print(DATA_PATH)

if os.path.exists(DATA_PATH):
    print("Dataset file found successfully!")
else:
    print("ERROR: Dataset file not found!")

file_size_bytes = os.path.getsize(DATA_PATH)

file_size_mb = file_size_bytes / (1024 ** 2)

print("Dataset file size:", round(file_size_mb, 2), "MB")
```

```python
with zipfile.ZipFile(DATA_PATH, 'r') as zip_ref:

    # Get the list of files inside the archive
    archive_files = zip_ref.namelist()

    # Display the archive contents
    print("Files inside the ZIP archive:\n")

    for file in archive_files:
        print(file)

    # Identify CSV files
    csv_files = [
        file for file in archive_files
        if file.lower().endswith('.csv')
    ]

    # Identify README files
    readme_files = [
        file for file in archive_files
        if 'readme' in os.path.basename(file).lower()
    ]

    print("\nNumber of CSV files:", len(csv_files))
    print("Number of README files:", len(readme_files))
```

```python
if len(csv_files) == 0:
    raise FileNotFoundError("No CSV file found in the ZIP archive.")

if len(readme_files) == 0:
    raise FileNotFoundError("No README file found in the ZIP archive.")

# Store the identified file names
CSV_NAME = csv_files[0]
README_NAME = readme_files[0]

print("CSV Dataset File:", CSV_NAME)
print("README Documentation File:", README_NAME)
```

#Data Exploration

##Inspect Dataset Columns

```python
# @title

with zipfile.ZipFile(DATA_PATH, 'r') as zip_ref:

    # Open the CSV file directly inside the ZIP archive
    with zip_ref.open(CSV_NAME) as csv_file:

        # Read only the CSV header without loading data rows
        df_header = pd.read_csv(csv_file, nrows=0)

# Extract column names
column_names = df_header.columns.tolist()

# Display dataset structure
print("Number of Columns:", len(column_names))

print("\nDataset Column Names:\n")

for index, column in enumerate(column_names, start=1):
    print(f"{index}. {column}")
```
