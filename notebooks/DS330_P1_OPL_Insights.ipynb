#Import Required Libraries

```python
import os
import zipfile

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display
from IPython.display import display, HTML

from google.colab import drive
drive.mount('/content/drive')
```

#Mount Google Drive & Define Dataset Path

```python
DATA_PATH = "/content/drive/MyDrive/DS330_Project/openpowerlifting-latest.zip"

print("Dataset path:")
print(DATA_PATH)

if os.path.exists(DATA_PATH):
    print("Dataset file found successfully!")
else:
    print("ERROR: Dataset file not found!")

file_size_bytes = os.path.getsize(DATA_PATH)

file_size_mb = file_size_bytes / (1024 ** 2)

print("Dataset file size:", round(file_size_mb, 2), "MB")
```

```python
with zipfile.ZipFile(DATA_PATH, 'r') as zip_ref:

    # Get all file names inside the ZIP archive
    archive_files = zip_ref.namelist()

    print("Files inside the ZIP archive:\n")

    for file in archive_files:
        print(file)

    # Identify CSV files only
    csv_files = [
        file for file in archive_files
        if file.lower().endswith('.csv')
    ]
```

```python
if len(csv_files) == 0:
    raise FileNotFoundError("No CSV file found in the ZIP archive.")

CSV_NAME = csv_files[0]

print("CSV Dataset File:", CSV_NAME)
```

#Data Exploration

```python
# @title Dataset Structure
with zipfile.ZipFile(DATA_PATH, 'r') as zip_ref:

    # Open the CSV file directly inside the ZIP archive
    with zip_ref.open(CSV_NAME) as csv_file:

        # Read only the CSV header without loading data rows
        df_header = pd.read_csv(csv_file, nrows=0)

# Extract column names
column_names = df_header.columns.tolist()

# Display section heading and description
display(HTML("""
<div style="
    border-left:3px solid #E23B43;
    padding:5px 12px;
    margin-bottom:15px;
    font-family:Arial;
">
    <h2 style="color:#222222; margin:0 0 12px 0;">
        Dataset Structure
    </h2>

    <p style="color:#444444; font-size:14px; margin:0;">
        An overview of the OpenPowerlifting dataset structure,
        including the total number of features and their column names.
    </p>
</div>
"""))

# Display number of columns
display(HTML(f"""
<div style="
    background:#F2F2F2;
    border-left:4px solid #E23B43;
    padding:12px 16px;
    margin-bottom:15px;
    font-family:Arial;
">
    <span style="color:#444;font-size:13px;">
        TOTAL NUMBER OF COLUMNS
    </span>

    <h2 style="color:#E23B43;margin:5px 0 0 0;">
        {len(column_names)}
    </h2>
</div>
"""))

# Create a structured table
columns_df = pd.DataFrame({
    "No.": range(1, len(column_names) + 1),
    "Column Name": column_names
})

# Apply professional table styling
styled_columns = (
    columns_df.style
    .hide(axis="index")
    .set_properties(**{
        'text-align': 'left',
        'padding': '8px',
        'font-size': '13px',
        'color': '#333333',
        'border': '1px solid #dddddd'
    })
    .set_table_styles([
        {
            'selector': 'th',
            'props': [
                ('background-color', '#E8E8E8'),
                ('color', '#222222'),
                ('font-weight', 'bold'),
                ('text-align', 'left'),
                ('padding', '9px'),
                ('border', '1px solid #cccccc')
            ]
        },
        {
            'selector': 'tbody tr:nth-child(even)',
            'props': [('background-color', '#F0F0F0')]
        },
        {
            'selector': 'tbody tr:nth-child(odd)',
            'props': [('background-color', '#FFFFFF')]
        },
        {
            'selector': 'table',
            'props': [
                ('border-collapse', 'collapse'),
                ('width', '100%'),
                ('font-family', 'Arial')
            ]
        }
    ])
)

display(styled_columns)
```
