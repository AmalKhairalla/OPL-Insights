## Decisions (fixed before coding)

| Decision | Value | Reason |
|---|---|---|
| Target | TotalKg | Continuous, many distinct values |
| Event filter | SBD only | Total means the same thing in every row |
| Features | Age, BodyweightKg, Year, Age_missing, Tested, Sex, Equipment | No leakage |
| Excluded | Federation, Country | Too many categories; project is about training |
| Split | Grouped by Name, 70/15/15, seed 42 | Same lifter has many similar rows; random split leaks |
| Sample | 200,000 rows, seed 42 (before split) | Course requirement |
| Epochs / seed | 30 / 42 in every run | Fair comparison |
| Target scaling | Standardized for training; MSE reported in kg² | Stable training, comparable numbers |

Leakage columns removed: Best3*Kg, Squat1-4Kg, Bench1-4Kg, Deadlift1-4Kg, Dots, Wilks, Glossbrenner, Goodlift, Place.

#The imports, seed, and device

```python
import os
import glob
import zipfile
import time
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GroupShuffleSplit

SEED = 42

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
```

```python
USE_COLS = ['Name', 'Sex', 'Event', 'Equipment', 'Age',
            'BodyweightKg', 'TotalKg', 'Tested', 'Date']

DTYPES = {
    'Sex': 'category',
    'Event': 'category',
    'Equipment': 'category',
    'Tested': 'category',
    'Age': 'float32',
    'BodyweightKg': 'float32',
    'TotalKg': 'float32',
}

df = pd.read_csv(CSV_PATH, usecols=USE_COLS, dtype=DTYPES)

print('Shape:', df.shape)
print(f'Memory: {df.memory_usage(deep=True).sum() / 1e6:.0f} MB')
display(df.head())
print(df.dtypes)
print(df.isna().mean().round(3))
```

#SBD filtration + cleaning log

```python
clean_log = []

def log_step(description, before, after):
    clean_log.append({'Step': description, 'Rows before': before,
                      'Rows after': after, 'Removed': before - after})
    print(f'{description}: {before:,} -> {after:,}  (removed {before - after:,})')

n0 = len(df)
df = df[df['Event'] == 'SBD'].copy()
log_step("Keep Event == 'SBD'", n0, len(df))

df = df.drop(columns='Event')
print(df.shape)
```

#The Target

```python
n0 = len(df)
df = df.dropna(subset=['TotalKg'])
log_step('Drop missing TotalKg', n0, len(df))

n0 = len(df)
df = df[df['TotalKg'] > 0]
log_step('Drop TotalKg <= 0', n0, len(df))
```

#Weight, gender, equipment

```python
n0 = len(df)
df = df.dropna(subset=['BodyweightKg'])
log_step('Drop missing BodyweightKg', n0, len(df))

print(df['Sex'].value_counts())
print()
print(df['Equipment'].value_counts())
```

#Tested and Date

```python
df['Tested'] = (df['Tested'] == 'Yes').astype('int8')

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Year'] = df['Date'].dt.year

n0 = len(df)
df = df.dropna(subset=['Year'])
log_step('Drop rows with unparsable Date', n0, len(df))

df['Year'] = df['Year'].astype('int16')
df = df.drop(columns='Date')

print(df['Tested'].value_counts())
print(df['Year'].describe())
```

#Treatment after the split

##Age

```python
age_median = train_df['Age'].median()
print('Age median (from train):', age_median)

for d in (train_df, val_df, test_df):
    d['Age_missing'] = d['Age'].isna().astype('int8')
    d['Age'] = d['Age'].fillna(age_median)
```

##One-hot

```python
CAT_COLS = ['Sex', 'Equipment']

encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoder.fit(train_df[CAT_COLS].astype(str))
cat_names = encoder.get_feature_names_out(CAT_COLS)
print(list(cat_names))

def encode(d):
    arr = encoder.transform(d[CAT_COLS].astype(str))
    return pd.DataFrame(arr, columns=cat_names, index=d.index)
```

##Standardization and X-Building

```python
NUM_COLS = ['Age', 'BodyweightKg', 'Year']
BIN_COLS = ['Tested', 'Age_missing']

scaler = StandardScaler().fit(train_df[NUM_COLS])

def build_X(d):
    num = pd.DataFrame(scaler.transform(d[NUM_COLS]), columns=NUM_COLS, index=d.index)
    binary = d[BIN_COLS].astype('float32')
    return pd.concat([num, binary, encode(d)], axis=1)

X_train = build_X(train_df)
X_val = build_X(val_df)
X_test = build_X(test_df)

assert list(X_train.columns) == list(X_val.columns) == list(X_test.columns)
print('Features:', list(X_train.columns))
print('Train numeric means (should be ~0):')
print(X_train[NUM_COLS].mean().round(3))
print(X_train[NUM_COLS].std().round(3))
```

##The Target

```python
y_mean = train_df['TotalKg'].mean()
y_std = train_df['TotalKg'].std()
print(f'y_mean={y_mean:.2f}  y_std={y_std:.2f}')

def scale_y(d):
    return ((d['TotalKg'] - y_mean) / y_std).values.astype('float32')

y_train = scale_y(train_df)
y_val = scale_y(val_df)
y_test = scale_y(test_df)
```

#Switching to PyTorch

##Tensors

```python
def to_tensor(arr):
    return torch.tensor(np.asarray(arr, dtype=np.float32)).to(device)

X_train_t = to_tensor(X_train.values)
X_val_t = to_tensor(X_val.values)
X_test_t = to_tensor(X_test.values)

y_train_t = to_tensor(y_train).view(-1, 1)
y_val_t = to_tensor(y_val).view(-1, 1)
y_test_t = to_tensor(y_test).view(-1, 1)

n_features = X_train_t.shape[1]
n_train = X_train_t.shape[0]

print('X_train:', X_train_t.shape, ' y_train:', y_train_t.shape)
print('X_val  :', X_val_t.shape, ' y_val  :', y_val_t.shape)
print('X_test :', X_test_t.shape, ' y_test :', y_test_t.shape)
print('n_features =', n_features)
```

The division into batches happens inside the train_model (manual with shuffle in each epoch), because the batch_size changes between experiments.

#Building the model

```python
def build_model(n_features, hidden_layers):
    layers = []
    in_dim = n_features
    for h in hidden_layers:
        layers.append(nn.Linear(in_dim, h))
        layers.append(nn.ReLU())
        in_dim = h
    layers.append(nn.Linear(in_dim, 1))
    return nn.Sequential(*layers).to(device)
```

##Training function

```python
def get_optimizer(name, params, lr, momentum):
    if name == 'sgd':
        return torch.optim.SGD(params, lr=lr)
    if name == 'sgd_momentum':
        return torch.optim.SGD(params, lr=lr, momentum=momentum)
    if name == 'adam':
        return torch.optim.Adam(params, lr=lr)
    raise ValueError(f'Unknown optimizer: {name}')

def get_loss(name):
    return {'mse': nn.MSELoss(),
            'l1': nn.L1Loss(),
            'huber': nn.HuberLoss()}[name]

@torch.no_grad()
def eval_mse(model, X, y):
    model.eval()
    mse_scaled = ((model(X) - y) ** 2).mean().item()
    return mse_scaled * (y_std ** 2)

def train_model(hidden_layers, optimizer_name='sgd', lr=0.01, momentum=0.0,
                batch_size=64, loss_name='mse', epochs=30, seed=SEED, verbose=True):
    set_seed(seed)
    model = build_model(n_features, hidden_layers)
    optimizer = get_optimizer(optimizer_name, model.parameters(), lr, momentum)
    criterion = get_loss(loss_name)
    g = torch.Generator()
    g.manual_seed(seed)

    train_losses, val_losses = [], []
    diverged = False
    start = time.time()

    for epoch in range(epochs):
        model.train()
        perm = torch.randperm(n_train, generator=g).to(device)
        for i in range(0, n_train, batch_size):
            idx = perm[i:i + batch_size]
            xb, yb = X_train_t[idx], y_train_t[idx]
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)
            loss.backward()
            optimizer.step()

        tr = eval_mse(model, X_train_t, y_train_t)
        va = eval_mse(model, X_val_t, y_val_t)
        train_losses.append(tr)
        val_losses.append(va)

        if verbose and (epoch == 0 or (epoch + 1) % 5 == 0):
            print(f'epoch {epoch + 1:2d}  train MSE {tr:9.2f}  val MSE {va:9.2f}')

        if not (np.isfinite(tr) and np.isfinite(va)):
            diverged = True
            print(f'Diverged at epoch {epoch + 1} (loss is nan/inf)')
            break

    elapsed = time.time() - start
    return {'train_losses': train_losses, 'val_losses': val_losses,
            'time': elapsed, 'model': model, 'diverged': diverged}
```

##Auxiliary functions

```python
RESULTS_CSV = '/content/drive/MyDrive/DS330_Project/results_baseline.csv'
results = []

def log_run(run_id, topic, setting, res, observation=''):
    results.append({
        'Run': run_id,
        'Topic': topic,
        'Setting changed from baseline': setting,
        'Train MSE': round(res['train_losses'][-1], 2),
        'Val MSE': round(res['val_losses'][-1], 2),
        'Test MSE': '—',
        'Time (s)': round(res['time'], 1),
        'Observation': observation,
    })
    pd.DataFrame(results).to_csv(RESULTS_CSV, index=False)
    return pd.DataFrame(results)

def plot_losses(res, linear_val_mse=None, title=''):
    ep = range(1, len(res['train_losses']) + 1)
    plt.figure(figsize=(8, 4.5))
    plt.plot(ep, res['train_losses'], color='gray', marker='o', ms=3, label='Training')
    plt.plot(ep, res['val_losses'], color='tab:blue', marker='o', ms=3, label='Validation')
    if linear_val_mse is not None:
        plt.axhline(linear_val_mse, color='black', ls='--', lw=1,
                    label=f'Linear regression (validation): {linear_val_mse:.1f}')
    plt.xlabel('Epoch')
    plt.ylabel('MSE (kg²)')
    plt.title(title)
    plt.legend()
    plt.show()
```

#Baseline

##Linear regression (PyTorch)

```python
res_linear = train_model(hidden_layers=[], optimizer_name='sgd',
                         lr=0.01, batch_size=64, epochs=30)
linear_val_mse = res_linear['val_losses'][-1]

log_run(1, 'Baseline', 'Linear regression (no hidden layers), SGD lr 0.01, batch 64',
        res_linear, 'Reference point')
```

##Sanity check with Sklearn

```python
y_train_raw = train_df['TotalKg'].values
y_val_raw = val_df['TotalKg'].values

sk = LinearRegression().fit(X_train.values, y_train_raw)
sk_train_mse = np.mean((sk.predict(X_train.values) - y_train_raw) ** 2)
sk_val_mse = np.mean((sk.predict(X_val.values) - y_val_raw) ** 2)

print(f'sklearn  : train {sk_train_mse:.2f}  val {sk_val_mse:.2f}')
print(f'PyTorch  : train {res_linear["train_losses"][-1]:.2f}  val {linear_val_mse:.2f}')
```

##The basic NN

```python
res_nn = train_model(hidden_layers=[64, 64, 64], optimizer_name='sgd',
                     lr=0.01, batch_size=64, epochs=30)

log_run(2, 'Baseline', 'NN [64, 64, 64], SGD, lr 0.01, batch 64', res_nn, '')
```

##Drawing and comparison

```python
plot_losses(res_nn, linear_val_mse, 'Baseline NN [64,64,64], SGD, lr 0.01, batch 64')

nn_val_mse = res_nn['val_losses'][-1]
improvement = (linear_val_mse - nn_val_mse) / linear_val_mse * 100
print(f'Linear val MSE: {linear_val_mse:.2f}')
print(f'NN val MSE    : {nn_val_mse:.2f}')
print(f'NN is {improvement:.1f}% better than linear regression on validation')
display(pd.DataFrame(results))
```
