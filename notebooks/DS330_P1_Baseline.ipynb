##Data preparation

```python

# Normalize names for grouping
df = df.reset_index(drop=True).copy()

groups = (
    df["Name"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

assert groups.notna().all()
assert groups.ne("").all()

# First reserve test groups
split_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=42
)

trainval_idx, test_idx = next(
    split_test.split(df, groups=groups)
)

# Split remaining groups into train and validation
split_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15 / 0.85,
    random_state=43
)

train_rel, val_rel = next(
    split_val.split(
        df.iloc[trainval_idx],
        groups=groups.iloc[trainval_idx]
    )
)

train_idx = trainval_idx[train_rel]
val_idx = trainval_idx[val_rel]

train_df = df.iloc[train_idx].copy()
val_df = df.iloc[val_idx].copy()
test_df = df.iloc[test_idx].copy()

# Verify zero name overlap
train_names = set(groups.iloc[train_idx])
val_names = set(groups.iloc[val_idx])
test_names = set(groups.iloc[test_idx])

assert train_names.isdisjoint(val_names)
assert train_names.isdisjoint(test_names)
assert val_names.isdisjoint(test_names)

print("No recorded-name overlap!")

for name, frame in [
    ("Training", train_df),
    ("Validation", val_df),
    ("Test", test_df)
]:
    print(
        name,
        len(frame),
        f"({len(frame)/len(df)*100:.2f}%)"
    )
```

```python

numeric_features = [
    "Age",
    "BodyweightKg",
    "Best3SquatKg",
    "Best3BenchKg"
]

categorical_features = ["Sex"]

target = "Best3DeadliftKg"

features = numeric_features + categorical_features

# Separate predictors and target
X_train = train_df[features]
X_val = val_df[features]

y_train = train_df[target]
y_val = val_df[target]

# Numeric preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Numeric + categorical preprocessing
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("sex", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ), categorical_features)
])

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Reuse training transformations
X_val_processed = preprocessor.transform(X_val)

print("Training shape:", X_train_processed.shape)
print("Validation shape:", X_val_processed.shape)

print("\nEncoded feature names:")
print(preprocessor.get_feature_names_out())
```

```python

import pandas as pd

# Use the existing split dataframes
# No need to redo splitting or preprocessing

split_data = df.copy()

# Verify every record belongs to exactly one split
assert train_df.index.is_unique
assert val_df.index.is_unique
assert test_df.index.is_unique

assert set(train_df.index).isdisjoint(val_df.index)
assert set(train_df.index).isdisjoint(test_df.index)
assert set(val_df.index).isdisjoint(test_df.index)

assert (
    set(train_df.index)
    | set(val_df.index)
    | set(test_df.index)
) == set(split_data.index)

# Attach labels to existing records
split_data["Split"] = ""

split_data.loc[train_df.index, "Split"] = "train"
split_data.loc[val_df.index, "Split"] = "validation"
split_data.loc[test_df.index, "Split"] = "test"

# Save the fixed dataset
file_path = "/content/FPR_SinglePly_fixed_split.csv.gz"

split_data.to_csv(
    file_path,
    index=False,
    compression="gzip"
)

print("Saved successfully!")
print(split_data["Split"].value_counts())
```

###PyTorch Neural Network Baseline

```python

import torch
import numpy as np
import random

from torch import nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import StandardScaler

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# Standardize target using training data only
y_scaler = StandardScaler()

y_train_scaled = y_scaler.fit_transform(
    np.asarray(y_train).reshape(-1, 1)
)

# Convert inputs to tensors
Xtr = torch.tensor(
    X_train_processed,
    dtype=torch.float32
).to(device)

Xva = torch.tensor(
    X_val_processed,
    dtype=torch.float32
).to(device)

ytr = torch.tensor(
    y_train_scaled,
    dtype=torch.float32
).to(device)

train_loader = DataLoader(
    TensorDataset(Xtr, ytr),
    batch_size=64,
    shuffle=True
)

print("Input features:", Xtr.shape[1])
print("Device:", device)
```

```python
import time
import copy
import random
import numpy as np
import torch
from torch import nn
from torch.utils.data import TensorDataset, DataLoader

# Targets in original kg scale (numpy) + target scaler statistics
y_train_np = np.asarray(y_train, dtype=np.float64)
y_val_np = np.asarray(y_val, dtype=np.float64)
y_mean = float(y_scaler.mean_[0])
y_std = float(y_scaler.scale_[0])


def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)


def build_mlp(in_features, hidden):
    layers = []
    prev = in_features
    for h in hidden:
        layers += [nn.Linear(prev, h), nn.ReLU()]
        prev = h
    layers.append(nn.Linear(prev, 1))
    return nn.Sequential(*layers)


def make_optimizer(name, params, lr, momentum=0.0):
    name = name.lower()
    if name == "sgd":
        return torch.optim.SGD(params, lr=lr, momentum=momentum)
    if name == "adam":
        return torch.optim.Adam(params, lr=lr)
    if name == "adamw":
        return torch.optim.AdamW(params, lr=lr)
    raise ValueError(f"Unknown optimizer: {name}")


def make_loss(name, huber_delta=1.0):
    name = name.lower()
    if name == "mse":
        return nn.MSELoss()
    if name == "l1":
        return nn.L1Loss()
    if name == "huber":
        return nn.HuberLoss(delta=huber_delta)
    raise ValueError(f"Unknown loss: {name}")


def mse_in_kg(scaled_output, y_true_kg):
    # Convert predictions back to kg, then compute MSE (kg^2)
    pred_kg = scaled_output.cpu().numpy().astype(np.float64).ravel() * y_std + y_mean
    return float(np.mean((y_true_kg - pred_kg) ** 2))


def run_experiment(hidden=(64, 64, 64), optimizer="sgd", lr=0.01,
                   batch_size=64, loss="mse", momentum=0.0,
                   huber_delta=1.0, epochs=30, seed=42, verbose=True,
                   patience=None, min_delta=0.0):
    """Defaults = the baseline. Change ONE argument per experiment.
    patience=None -> fixed epochs (no early stopping).
    patience=k    -> stop if Val MSE does not improve for k epochs,
                     then restore the best epoch (chosen on validation only)."""
    set_seed(seed)

    model = build_mlp(Xtr.shape[1], hidden).to(device)
    opt = make_optimizer(optimizer, model.parameters(), lr, momentum)
    criterion = make_loss(loss, huber_delta)
    loader = DataLoader(TensorDataset(Xtr, ytr),
                        batch_size=batch_size, shuffle=True)

    train_losses, val_losses = [], []

    best_val = float("inf")
    best_epoch = 0
    best_state = None
    wait = 0
    stopped_early = False

    start = time.perf_counter()

    for epoch in range(epochs):
        model.train()
        for xb, yb in loader:
            opt.zero_grad()
            batch_loss = criterion(model(xb), yb)
            batch_loss.backward()
            opt.step()

        model.eval()
        with torch.no_grad():
            tr_mse = mse_in_kg(model(Xtr), y_train_np)
            va_mse = mse_in_kg(model(Xva), y_val_np)

        train_losses.append(tr_mse)
        val_losses.append(va_mse)

        if verbose:
            print(f"Epoch {epoch+1:02d}/{epochs} | "
                  f"Train MSE: {tr_mse:.2f} | Val MSE: {va_mse:.2f}")

        if not np.isfinite(va_mse):
            print("Run diverged (NaN/inf) -> stopping this run.")
            break

        # Track the best epoch (validation only)
        if va_mse < best_val - min_delta:
            best_val = va_mse
            best_epoch = epoch + 1
            wait = 0
            if patience is not None:
                best_state = copy.deepcopy(model.state_dict())
        else:
            wait += 1
            if patience is not None and wait >= patience:
                stopped_early = True
                print(f"Early stopping at epoch {epoch+1} "
                      f"(best epoch: {best_epoch}, best Val MSE: {best_val:.2f})")
                break

    elapsed = time.perf_counter() - start

    # With early stopping, go back to the best epoch's weights
    if patience is not None and best_state is not None:
        model.load_state_dict(best_state)
        final_idx = best_epoch - 1
    else:
        final_idx = -1

    return {
        "train_mse": train_losses[final_idx],
        "val_mse": val_losses[final_idx],
        "val_mse_last5": float(np.mean(val_losses[-5:])),
        "best_epoch": best_epoch,
        "best_val_mse": best_val,
        "stopped_early": stopped_early,
        "time": elapsed,
        "epochs_run": len(val_losses),
        "train_losses": train_losses,
        "val_losses": val_losses,
        "model": model,
    }

baseline_res = run_experiment()   # all defaults = the baseline

nn_train_losses = baseline_res["train_losses"]
nn_val_losses = baseline_res["val_losses"]
nn_time = baseline_res["time"]

print("\nBaseline NN completed!")
print(f"Final Train MSE: {baseline_res['train_mse']:.2f}")
print(f"Final Val MSE: {baseline_res['val_mse']:.2f}")
print(f"Val MSE (mean of last 5 epochs): {baseline_res['val_mse_last5']:.2f}")
print(f"Training time: {nn_time:.2f} seconds")
```

```python
import time
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

pd.set_option("display.max_colwidth", 60)
pd.set_option("display.width", 250)

# Baseline configuration (every experiment changes ONE thing from this)
BASE = dict(hidden=(64, 64, 64), optimizer="sgd", lr=0.01,
            batch_size=64, loss="mse", momentum=0.0, huber_delta=1.0)

experiments = [
    # ---- Learning rate (6) ----
    ("Learning rate", "lr = 0.0003", dict(lr=0.0003)),
    ("Learning rate", "lr = 0.001",  dict(lr=0.001)),
    ("Learning rate", "lr = 0.003",  dict(lr=0.003)),
    ("Learning rate", "lr = 0.03",   dict(lr=0.03)),
    ("Learning rate", "lr = 0.1",    dict(lr=0.1)),
    ("Learning rate", "lr = 0.3",    dict(lr=0.3)),
    # ---- Optimizer (5) ----
    ("Optimizer", "SGD + momentum 0.5", dict(momentum=0.5)),
    ("Optimizer", "SGD + momentum 0.9", dict(momentum=0.9)),
    ("Optimizer", "Adam (lr 0.01)", dict(optimizer="adam")),
    ("Optimizer", "Adam (lr 0.001 = its default; 2 changes)", dict(optimizer="adam", lr=0.001)),
    ("Optimizer", "AdamW (lr 0.01)", dict(optimizer="adamw")),
    # ---- Batch size (5) ----
    ("Batch size", "batch 16",   dict(batch_size=16)),
    ("Batch size", "batch 32",   dict(batch_size=32)),
    ("Batch size", "batch 128",  dict(batch_size=128)),
    ("Batch size", "batch 256",  dict(batch_size=256)),
    ("Batch size", "batch 1024", dict(batch_size=1024)),
    # ---- Depth and width (8) ----
    ("Depth/Width", "1 hidden layer [64]",    dict(hidden=(64,))),
    ("Depth/Width", "2 hidden layers [64]x2", dict(hidden=(64,) * 2)),
    ("Depth/Width", "4 hidden layers [64]x4", dict(hidden=(64,) * 4)),
    ("Depth/Width", "6 hidden layers [64]x6", dict(hidden=(64,) * 6)),
    ("Depth/Width", "width 16  [16]x3",  dict(hidden=(16,) * 3)),
    ("Depth/Width", "width 32  [32]x3",  dict(hidden=(32,) * 3)),
    ("Depth/Width", "width 128 [128]x3", dict(hidden=(128,) * 3)),
    ("Depth/Width", "width 256 [256]x3", dict(hidden=(256,) * 3)),
    # ---- Loss function (4) ----
    ("Loss function", "L1 loss", dict(loss="l1")),
    ("Loss function", "Huber (delta 0.1)", dict(loss="huber", huber_delta=0.1)),
    ("Loss function", "Huber (delta 0.5)", dict(loss="huber", huber_delta=0.5)),
    ("Loss function", "Huber (delta 1.0)", dict(loss="huber", huber_delta=1.0)),
]
print(len(experiments) + 2, "experiments in total (incl. linear + baseline)")


def count_changes(cfg):
    """How many settings differ from the baseline (should be 1)."""
    n = 0
    for k, v in cfg.items():
        if k == "huber_delta" and cfg.get("loss", "mse") == "huber":
            continue
        if BASE[k] != v:
            n += 1
    return n


all_runs = {}   # run_id -> info (kept if you re-run this cell)


def add_run(run_id, topic, setting, cfg, res):
    all_runs[run_id] = dict(topic=topic, setting=setting, cfg=cfg,
                            changes=count_changes(cfg), res=res)


# Run 1: Linear regression (already trained above, we only time the fit)
t0 = time.perf_counter()
LinearRegression().fit(X_train_processed, y_train)
linear_time = time.perf_counter() - t0
linear_res = dict(train_mse=linear_train_mse, val_mse=linear_val_mse,
                  val_mse_last5=linear_val_mse, best_epoch=np.nan,
                  time=linear_time, train_losses=[], val_losses=[])
add_run(1, "Baseline", "Linear regression (no hidden layers)", BASE, linear_res)

# Run 2: Baseline neural network (already trained above)
add_run(2, "Baseline", "NN [64,64,64], SGD, lr 0.01, batch 64", BASE, baseline_res)

# Runs 3..30
for run_id, (topic, label, override) in enumerate(experiments, start=3):
    if run_id in all_runs:
        continue
    cfg = {**BASE, **override}
    print(f"Run {run_id}: {topic} | {label}")
    try:
        res = run_experiment(**cfg, verbose=False)
    except Exception as e:
        print("   FAILED:", e)
        continue
    add_run(run_id, topic, label, cfg, res)
    print(f"   Train {res['train_mse']:.2f} | Val {res['val_mse']:.2f} | {res['time']:.0f}s")
```

```python
# Write your own observation for each run AFTER you look at the results,
# e.g. observations = {3: "Too slow in 30 epochs", 7: "Diverged"}
observations = {}


def build_table():
    rows = []
    for run_id in sorted(all_runs):
        d = all_runs[run_id]
        r = d["res"]
        rows.append({
            "Run": run_id,
            "Topic": d["topic"],
            "Setting changed from baseline": d["setting"],
            "Changes": 0 if run_id <= 2 else d["changes"],
            "Train MSE": r["train_mse"],
            "Val MSE": r["val_mse"],
            "Val MSE (last-5 avg)": r["val_mse_last5"],
            "Best epoch": r["best_epoch"],
            "Test MSE": "—",
            "Time (s)": r["time"],
        })
    table = pd.DataFrame(rows)
    table["Improvement vs Linear (%)"] = (
        (linear_val_mse - table["Val MSE"]) / linear_val_mse * 100
    )
    table["Observation"] = table["Run"].map(observations).fillna("")
    return table


results_df = build_table()
display(results_df.round(2))
results_df.to_csv("experiment_results.csv", index=False)
```
