##Data preparation

```python

# Normalize names for grouping
df = df.reset_index(drop=True).copy()

groups = (
    df["Name"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

assert groups.notna().all()
assert groups.ne("").all()

# First reserve test groups
split_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=42
)

trainval_idx, test_idx = next(
    split_test.split(df, groups=groups)
)

# Split remaining groups into train and validation
split_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15 / 0.85,
    random_state=43
)

train_rel, val_rel = next(
    split_val.split(
        df.iloc[trainval_idx],
        groups=groups.iloc[trainval_idx]
    )
)

train_idx = trainval_idx[train_rel]
val_idx = trainval_idx[val_rel]

train_df = df.iloc[train_idx].copy()
val_df = df.iloc[val_idx].copy()
test_df = df.iloc[test_idx].copy()

# Verify zero name overlap
train_names = set(groups.iloc[train_idx])
val_names = set(groups.iloc[val_idx])
test_names = set(groups.iloc[test_idx])

assert train_names.isdisjoint(val_names)
assert train_names.isdisjoint(test_names)
assert val_names.isdisjoint(test_names)

print("No recorded-name overlap!")

for name, frame in [
    ("Training", train_df),
    ("Validation", val_df),
    ("Test", test_df)
]:
    print(
        name,
        len(frame),
        f"({len(frame)/len(df)*100:.2f}%)"
    )
```

```python

numeric_features = [
    "Age",
    "BodyweightKg",
    "Best3SquatKg",
    "Best3BenchKg"
]

categorical_features = ["Sex"]

target = "Best3DeadliftKg"

features = numeric_features + categorical_features

# Separate predictors and target
X_train = train_df[features]
X_val = val_df[features]

y_train = train_df[target]
y_val = val_df[target]

# Numeric preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Numeric + categorical preprocessing
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("sex", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ), categorical_features)
])

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Reuse training transformations
X_val_processed = preprocessor.transform(X_val)

print("Training shape:", X_train_processed.shape)
print("Validation shape:", X_val_processed.shape)

print("\nEncoded feature names:")
print(preprocessor.get_feature_names_out())
```

```python

import pandas as pd

# Use the existing split dataframes
# No need to redo splitting or preprocessing

split_data = df.copy()

# Verify every record belongs to exactly one split
assert train_df.index.is_unique
assert val_df.index.is_unique
assert test_df.index.is_unique

assert set(train_df.index).isdisjoint(val_df.index)
assert set(train_df.index).isdisjoint(test_df.index)
assert set(val_df.index).isdisjoint(test_df.index)

assert (
    set(train_df.index)
    | set(val_df.index)
    | set(test_df.index)
) == set(split_data.index)

# Attach labels to existing records
split_data["Split"] = ""

split_data.loc[train_df.index, "Split"] = "train"
split_data.loc[val_df.index, "Split"] = "validation"
split_data.loc[test_df.index, "Split"] = "test"

# Save the fixed dataset
file_path = "/content/FPR_SinglePly_fixed_split.csv.gz"

split_data.to_csv(
    file_path,
    index=False,
    compression="gzip"
)

print("Saved successfully!")
print(split_data["Split"].value_counts())
```

##Baseline Models — Linear Regression vs. Deep Neural Network

###Linear Regression Baseline

###PyTorch Neural Network Baseline

```python

import torch
import numpy as np
import random

from torch import nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import StandardScaler

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# Standardize target using training data only
y_scaler = StandardScaler()

y_train_scaled = y_scaler.fit_transform(
    np.asarray(y_train).reshape(-1, 1)
)

# Convert inputs to tensors
Xtr = torch.tensor(
    X_train_processed,
    dtype=torch.float32
).to(device)

Xva = torch.tensor(
    X_val_processed,
    dtype=torch.float32
).to(device)

ytr = torch.tensor(
    y_train_scaled,
    dtype=torch.float32
).to(device)

train_loader = DataLoader(
    TensorDataset(Xtr, ytr),
    batch_size=64,
    shuffle=True
)

print("Input features:", Xtr.shape[1])
print("Device:", device)
```

```python
import time
import random
import numpy as np
import torch
from torch import nn
from torch.utils.data import TensorDataset, DataLoader

# Targets in original kg scale (numpy) + target scaler statistics
y_train_np = np.asarray(y_train, dtype=np.float64)
y_val_np = np.asarray(y_val, dtype=np.float64)
y_mean = float(y_scaler.mean_[0])
y_std = float(y_scaler.scale_[0])


def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)


def build_mlp(in_features, hidden):
    layers = []
    prev = in_features
    for h in hidden:
        layers += [nn.Linear(prev, h), nn.ReLU()]
        prev = h
    layers.append(nn.Linear(prev, 1))
    return nn.Sequential(*layers)


def make_optimizer(name, params, lr, momentum=0.0):
    name = name.lower()
    if name == "sgd":
        return torch.optim.SGD(params, lr=lr, momentum=momentum)
    if name == "adam":
        return torch.optim.Adam(params, lr=lr)
    if name == "adamw":
        return torch.optim.AdamW(params, lr=lr)
    raise ValueError(f"Unknown optimizer: {name}")


def make_loss(name, huber_delta=1.0):
    name = name.lower()
    if name == "mse":
        return nn.MSELoss()
    if name == "l1":
        return nn.L1Loss()
    if name == "huber":
        return nn.HuberLoss(delta=huber_delta)
    raise ValueError(f"Unknown loss: {name}")


def mse_in_kg(scaled_output, y_true_kg):
    # Convert predictions back to kg, then compute MSE (kg^2)
    pred_kg = scaled_output.cpu().numpy().astype(np.float64).ravel() * y_std + y_mean
    return float(np.mean((y_true_kg - pred_kg) ** 2))


def run_experiment(hidden=(64, 64, 64), optimizer="sgd", lr=0.01,
                   batch_size=64, loss="mse", momentum=0.0,
                   huber_delta=1.0, epochs=30, seed=42, verbose=True):
    """Defaults = the baseline. Change ONE argument per experiment."""
    set_seed(seed)

    model = build_mlp(Xtr.shape[1], hidden).to(device)
    opt = make_optimizer(optimizer, model.parameters(), lr, momentum)
    criterion = make_loss(loss, huber_delta)
    loader = DataLoader(TensorDataset(Xtr, ytr),
                        batch_size=batch_size, shuffle=True)

    train_losses, val_losses = [], []
    start = time.perf_counter()

    for epoch in range(epochs):
        model.train()
        for xb, yb in loader:
            opt.zero_grad()
            batch_loss = criterion(model(xb), yb)
            batch_loss.backward()
            opt.step()

        model.eval()
        with torch.no_grad():
            tr_mse = mse_in_kg(model(Xtr), y_train_np)
            va_mse = mse_in_kg(model(Xva), y_val_np)

        train_losses.append(tr_mse)
        val_losses.append(va_mse)

        if verbose:
            print(f"Epoch {epoch+1:02d}/{epochs} | "
                  f"Train MSE: {tr_mse:.2f} | Val MSE: {va_mse:.2f}")

        if not np.isfinite(va_mse):
            print("Run diverged (NaN/inf) -> stopping this run.")
            break

    elapsed = time.perf_counter() - start

    return {
        "train_mse": train_losses[-1],
        "val_mse": val_losses[-1],
        "val_mse_last5": float(np.mean(val_losses[-5:])),
        "time": elapsed,
        "epochs_run": len(val_losses),
        "train_losses": train_losses,
        "val_losses": val_losses,
    }

baseline_res = run_experiment()   # all defaults = the baseline

nn_train_losses = baseline_res["train_losses"]
nn_val_losses = baseline_res["val_losses"]
nn_time = baseline_res["time"]

print("\nBaseline NN completed!")
print(f"Final Train MSE: {baseline_res['train_mse']:.2f}")
print(f"Final Val MSE: {baseline_res['val_mse']:.2f}")
print(f"Val MSE (mean of last 5 epochs): {baseline_res['val_mse_last5']:.2f}")
print(f"Training time: {nn_time:.2f} seconds")
```
