#Data preparation

```python

# Normalize names for grouping
df = df.reset_index(drop=True).copy()

groups = (
    df["Name"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

assert groups.notna().all()
assert groups.ne("").all()

# First reserve test groups
split_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=42
)

trainval_idx, test_idx = next(
    split_test.split(df, groups=groups)
)

# Split remaining groups into train and validation
split_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15 / 0.85,
    random_state=43
)

train_rel, val_rel = next(
    split_val.split(
        df.iloc[trainval_idx],
        groups=groups.iloc[trainval_idx]
    )
)

train_idx = trainval_idx[train_rel]
val_idx = trainval_idx[val_rel]

train_df = df.iloc[train_idx].copy()
val_df = df.iloc[val_idx].copy()
test_df = df.iloc[test_idx].copy()

# Verify zero name overlap
train_names = set(groups.iloc[train_idx])
val_names = set(groups.iloc[val_idx])
test_names = set(groups.iloc[test_idx])

assert train_names.isdisjoint(val_names)
assert train_names.isdisjoint(test_names)
assert val_names.isdisjoint(test_names)

print("No recorded-name overlap!")

for name, frame in [
    ("Training", train_df),
    ("Validation", val_df),
    ("Test", test_df)
]:
    print(
        name,
        len(frame),
        f"({len(frame)/len(df)*100:.2f}%)"
    )
```

```python

numeric_features = [
    "Age",
    "BodyweightKg",
    "Best3SquatKg",
    "Best3BenchKg"
]

categorical_features = ["Sex"]

target = "Best3DeadliftKg"

features = numeric_features + categorical_features

# Separate predictors and target
X_train = train_df[features]
X_val = val_df[features]

y_train = train_df[target]
y_val = val_df[target]

# Numeric preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Numeric + categorical preprocessing
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("sex", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ), categorical_features)
])

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Reuse training transformations
X_val_processed = preprocessor.transform(X_val)

print("Training shape:", X_train_processed.shape)
print("Validation shape:", X_val_processed.shape)

print("\nEncoded feature names:")
print(preprocessor.get_feature_names_out())
```

```python

import pandas as pd

# Use the existing split dataframes
# No need to redo splitting or preprocessing

split_data = df.copy()

# Verify every record belongs to exactly one split
assert train_df.index.is_unique
assert val_df.index.is_unique
assert test_df.index.is_unique

assert set(train_df.index).isdisjoint(val_df.index)
assert set(train_df.index).isdisjoint(test_df.index)
assert set(val_df.index).isdisjoint(test_df.index)

assert (
    set(train_df.index)
    | set(val_df.index)
    | set(test_df.index)
) == set(split_data.index)

# Attach labels to existing records
split_data["Split"] = ""

split_data.loc[train_df.index, "Split"] = "train"
split_data.loc[val_df.index, "Split"] = "validation"
split_data.loc[test_df.index, "Split"] = "test"

# Save the fixed dataset
file_path = "/content/FPR_SinglePly_fixed_split.csv.gz"

split_data.to_csv(
    file_path,
    index=False,
    compression="gzip"
)

print("Saved successfully!")
print(split_data["Split"].value_counts())
```

```python

# Fix random seeds
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


set_seed(SEED)

# Scale target using training data only
y_scaler = StandardScaler()

y_train_np = np.asarray(y_train, dtype=np.float64)
y_val_np = np.asarray(y_val, dtype=np.float64)

y_train_scaled = y_scaler.fit_transform(
    y_train_np.reshape(-1, 1)
)

# Keep training tensors on CPU for DataLoader
Xtr = torch.tensor(
    X_train_processed,
    dtype=torch.float32
)

ytr = torch.tensor(
    y_train_scaled,
    dtype=torch.float32
)

# Evaluation tensors
Xtr_eval = Xtr.to(device)

Xva_eval = torch.tensor(
    X_val_processed,
    dtype=torch.float32
).to(device)

y_mean = float(y_scaler.mean_[0])
y_std = float(y_scaler.scale_[0])

print("Train shape:", Xtr.shape)
print("Validation shape:", Xva_eval.shape)
print("Target mean:", round(y_mean, 2))
print("Target standard deviation:", round(y_std, 2))
```

###Build models, optimizers and losses

```python

def build_mlp(input_size, hidden):
    layers = []
    previous = input_size

    for neurons in hidden:
        layers.append(nn.Linear(previous, neurons))
        layers.append(nn.ReLU())
        previous = neurons

    layers.append(nn.Linear(previous, 1))

    return nn.Sequential(*layers)


def make_optimizer(name, parameters, lr, momentum=0.0):
    name = name.lower()

    if name == "sgd":
        return torch.optim.SGD(
            parameters, lr=lr, momentum=momentum
        )

    elif name == "adam":
        return torch.optim.Adam(parameters, lr=lr)

    elif name == "adamw":
        return torch.optim.AdamW(parameters, lr=lr)

    else:
        raise ValueError(f"Unknown optimizer: {name}")


def make_loss(name, huber_delta=1.0):
    name = name.lower()

    if name == "mse":
        return nn.MSELoss()

    elif name == "l1":
        return nn.L1Loss()

    elif name == "huber":
        return nn.HuberLoss(delta=huber_delta)

    else:
        raise ValueError(f"Unknown loss: {name}")


def evaluate_mse(model, X, y_true_kg):
    model.eval()

    with torch.no_grad():
        pred_scaled = model(X).cpu().numpy().ravel()

    pred_kg = pred_scaled * y_std + y_mean

    return float(
        mean_squared_error(y_true_kg, pred_kg)
    )


print("Model, optimizer, and loss functions are ready!")
```

```python

import os

def save_json_safely(path, payload):
    """Write a JSON file, then replace the destination."""
    temp_path = path.with_suffix(".tmp")

    with open(temp_path, "w") as f:
        json.dump(payload, f, indent=2, allow_nan=False)

    os.replace(temp_path, path)


def experiment_saved(run_id):
    """Only skip a run if its successful results are saved."""
    metrics_path = RESULTS_DIR / f"{run_id}.json"
    history_path = HISTORY_DIR / f"{run_id}.json"
    model_path = MODEL_DIR / f"{run_id}.pt"

    if not all(p.exists() for p in [
        metrics_path, history_path, model_path
    ]):
        return False

    try:
        with open(metrics_path) as f:
            metadata = json.load(f)

        return (
            metadata.get("status") == "completed"
            and metadata.get("run_id") == run_id
        )
    except (OSError, ValueError):
        return False


def save_run(run_id, result, config):
    """Save metrics, curves and best checkpoint."""
    metrics_path = RESULTS_DIR / f"{run_id}.json"
    history_path = HISTORY_DIR / f"{run_id}.json"
    model_path = MODEL_DIR / f"{run_id}.pt"

    # Save the best model weights
    checkpoint = {
        "model_state_dict": result["best_state"],
        "config": config,
        "best_epoch": result["best_epoch"],
        "best_val_mse": result["best_val_mse"],
        "y_mean": y_mean,
        "y_std": y_std
    }

    temp_model_path = MODEL_DIR / f"{run_id}.tmp"
    torch.save(checkpoint, temp_model_path)
    os.replace(temp_model_path, model_path)

    # Save complete learning curves
    history = {
        "train_mse": result["train_losses"],
        "val_mse": result["val_losses"]
    }
    save_json_safely(history_path, history)

    # Save metadata last to mark the run complete
    metrics = {
        key: value
        for key, value in result.items()
        if key not in [
            "best_state",
            "train_losses",
            "val_losses"
        ]
    }

    metrics["run_id"] = run_id
    metrics["config"] = config

    save_json_safely(metrics_path, metrics)

    print(f"Saved {run_id} to Google Drive")


print("Automatic saving functions are ready!")
```

```python

def run_experiment(
    config,
    max_epochs=100,
    patience=10,
    min_delta=0.1,
    seed=42,
    verbose=False
):
    set_seed(seed)

    model = build_mlp(
        Xtr.shape[1],
        config["hidden"]
    ).to(device)

    optimizer = make_optimizer(
        config["optimizer"],
        model.parameters(),
        config["lr"],
        config["momentum"]
    )

    criterion = make_loss(
        config["loss"],
        config["huber_delta"]
    )

    loader = DataLoader(
        TensorDataset(Xtr, ytr),
        batch_size=config["batch_size"],
        shuffle=True
    )

    train_losses = []
    val_losses = []

    best_val = float("inf")
    best_epoch = 0
    best_state = None

    # Track meaningful improvements for early stopping
    patience_reference = float("inf")
    wait = 0

    status = "completed"
    stop_reason = "maximum_epochs"

    start = time.perf_counter()

    for epoch in range(1, max_epochs + 1):
        model.train()

        diverged = False

        for xb, yb in loader:
            xb = xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            predictions = model(xb)
            batch_loss = criterion(predictions, yb)

            if not torch.isfinite(batch_loss).item():
                diverged = True
                break

            batch_loss.backward()
            optimizer.step()

        if diverged:
            status = "failed"
            stop_reason = "non_finite_training_loss"
            break

        # MSE measured in original kg²
        train_mse = evaluate_mse(
            model, Xtr_eval, y_train_np
        )

        val_mse = evaluate_mse(
            model, Xva_eval, y_val_np
        )

        if not (
            np.isfinite(train_mse)
            and np.isfinite(val_mse)
        ):
            status = "failed"
            stop_reason = "non_finite_mse"
            break

        train_losses.append(train_mse)
        val_losses.append(val_mse)

        # Save the true best checkpoint
        if val_mse < best_val:
            best_val = val_mse
            best_epoch = epoch
            best_state = copy.deepcopy(
                model.state_dict()
            )

        # Early stopping uses meaningful improvement
        if val_mse < patience_reference - min_delta:
            patience_reference = val_mse
            wait = 0
        else:
            wait += 1

        if verbose:
            print(
                f"Epoch {epoch:03d} | "
                f"Train MSE: {train_mse:.2f} | "
                f"Val MSE: {val_mse:.2f} | "
                f"Wait: {wait}/{patience}"
            )

        if wait >= patience:
            stop_reason = "early_stopping"
            break

    elapsed = time.perf_counter() - start

    # Metrics of last successfully evaluated epoch
    final_train_mse = (
        train_losses[-1] if train_losses else None
    )
    final_val_mse = (
        val_losses[-1] if val_losses else None
    )

    # Selected checkpoint: lowest validation MSE
    selected_train_mse = (
        train_losses[best_epoch - 1]
        if best_epoch else None
    )

    selected_val_mse = (
        best_val if best_epoch else None
    )

    # Restore the selected checkpoint
    if best_state is not None:
        model.load_state_dict(best_state)
    else:
        status = "failed"

    return {
        "status": status,
        "stop_reason": stop_reason,
        "best_epoch": best_epoch,
        "epochs_run": len(val_losses),
        "train_mse": selected_train_mse,
        "val_mse": selected_val_mse,
        "best_val_mse": selected_val_mse,
        "final_train_mse": final_train_mse,
        "final_val_mse": final_val_mse,
        "time_seconds": float(elapsed),
        "train_losses": train_losses,
        "val_losses": val_losses,
        "best_state": best_state
    }


print("Early-stopping experiment function is ready!")
```

```python

# Settings allowed to change for each topic
ALLOWED_CHANGES = {
    "Learning rate": {"lr"},
    "Optimizer": {"optimizer", "momentum", "lr"},
    "Loss function": {"loss", "huber_delta"},
    "Batch size": {"batch_size"},
    "Depth": {"hidden"},
    "Width": {"hidden"}
}

run_ids = []

for run_id, topic, changes in all_experiments:
    assert run_id not in run_ids, f"Duplicate: {run_id}"
    run_ids.append(run_id)

    assert set(changes).issubset(ALLOWED_CHANGES[topic])

    cfg = {**BASE_CONFIG, **changes}

    # At least one baseline setting must differ
    assert any(
        cfg[key] != BASE_CONFIG[key]
        for key in BASE_CONFIG
    )

    # Optimizer comparison: exactly one optimizer choice
    if topic == "Optimizer":
        assert (
            ("optimizer" in changes) !=
            ("momentum" in changes)
        )

    # Huber delta belongs to the selected loss function
    if topic == "Loss function":
        assert "loss" in changes

    # Only depth OR width is varied
    if topic == "Depth":
        assert all(
            h == 64 for h in cfg["hidden"]
        )

    if topic == "Width":
        assert len(cfg["hidden"]) == 3
        assert len(set(cfg["hidden"])) == 1

print("All 30 experiment configurations passed!")
```

```python

# 5.3 Automatic experiment runner

def run_experiment_group(
    experiment_list,
    retry_failed=False
):
    records = []

    for run_id, topic, changes in experiment_list:
        config = copy.deepcopy(BASE_CONFIG)
        config.update(changes)

        # Save and compare the full experimental methodology
        saved_config = {
            **config,
            "max_epochs": EARLY_STOPPING["max_epochs"],
            "patience": EARLY_STOPPING["patience"],
            "min_delta": EARLY_STOPPING["min_delta"],
            "seed": SEED
        }

        metrics_path = RESULTS_DIR / f"{run_id}.json"

        # Check previously saved metadata
        previous = None
        if metrics_path.exists():
            with open(metrics_path) as f:
                previous = json.load(f)

            # Convert tuples to lists for JSON comparison
            expected = json.loads(json.dumps(saved_config))

            if previous.get("config") != expected:
                raise ValueError(
                    f"{run_id} has saved results from a "
                    "different configuration. Use a new "
                    "run ID or investigate before overwriting."
                )

        if experiment_saved(run_id):
            print(f"{run_id}: already completed — skipping")
            records.append(previous)
            continue

        if (
            previous is not None
            and previous.get("status") == "failed"
            and not retry_failed
        ):
            print(f"{run_id}: previously failed — skipping")
            records.append(previous)
            continue

        print(f"\n{'=' * 55}")
        print(f"Running {run_id}: {topic}")
        print("Changed settings:", changes)

        try:
            result = run_experiment(
                config=config,
                max_epochs=EARLY_STOPPING["max_epochs"],
                patience=EARLY_STOPPING["patience"],
                min_delta=EARLY_STOPPING["min_delta"],
                seed=SEED,
                verbose=False
            )

            if result["status"] == "completed":
                # Use existing saving function
                save_run(
                    run_id,
                    result,
                    saved_config
                )

                print(
                    f"Best validation MSE: "
                    f"{result['best_val_mse']:.2f}"
                )
                print("Best epoch:", result["best_epoch"])
                print("Epochs trained:", result["epochs_run"])
                print("Stop reason:", result["stop_reason"])

            else:
                # Save failed run information and history
                failure = {
                    "run_id": run_id,
                    "topic": topic,
                    "status": "failed",
                    "stop_reason": result["stop_reason"],
                    "config": saved_config,
                    "best_val_mse": result["best_val_mse"],
                    "best_epoch": result["best_epoch"],
                    "epochs_run": result["epochs_run"],
                    "time_seconds": result["time_seconds"]
                }

                save_json_safely(metrics_path, failure)

                save_json_safely(
                    HISTORY_DIR / f"{run_id}.json",
                    {
                        "train_mse": [
                            float(x) for x in result["train_losses"]
                        ],
                        "val_mse": [
                            float(x) for x in result["val_losses"]
                        ]
                    }
                )

                print(f"{run_id}: training diverged")

        except Exception as error:
            # Record unexpected errors as failures too
            failure = {
                "run_id": run_id,
                "topic": topic,
                "status": "failed",
                "stop_reason": "exception",
                "error": str(error),
                "config": saved_config
            }

            save_json_safely(metrics_path, failure)
            print(f"{run_id} FAILED: {error}")

        # Read back the persisted result
        with open(metrics_path) as f:
            records.append(json.load(f))

    return pd.DataFrame(records)


print("Experiment runner is ready!")
```

```python

# Plot Learning Curves for ALL Experiments

def plot_experiment(run_id):

    history_path = HISTORY_DIR / f"{run_id}.json"
    metrics_path = RESULTS_DIR / f"{run_id}.json"

    # Check if files exist
    if not history_path.exists():
        print(f"History not found: {run_id}")
        return

    with open(history_path, "r") as f:
        history = json.load(f)

    train_curve = history["train_mse"]
    val_curve = history["val_mse"]

    if not train_curve or not val_curve:
        print(f"Empty history: {run_id}")
        return

    # Load experiment metrics if available
    metrics = {}

    if metrics_path.exists():
        with open(metrics_path, "r") as f:
            metrics = json.load(f)

    epochs_train = range(1, len(train_curve) + 1)
    epochs_val = range(1, len(val_curve) + 1)

    plt.figure(figsize=(10, 5))

    # Training curve
    plt.plot(
        epochs_train,
        train_curve,
        label="Training MSE",
        linewidth=2
    )

    # Validation curve
    plt.plot(
        epochs_val,
        val_curve,
        label="Validation MSE",
        linewidth=2
    )

    # Linear Regression reference
    plt.axhline(
        linear_val_mse,
        linestyle="--",
        label=f"Linear Regression: {linear_val_mse:.2f}"
    )

    # Mark best validation epoch
    best_epoch = metrics.get("best_epoch", 0)
    best_val = metrics.get("best_val_mse")

    if best_epoch and best_val is not None:
        plt.scatter(
            best_epoch,
            best_val,
            marker="*",
            s=180,
            label=f"Best Epoch: {best_epoch}",
            zorder=5
        )

    plt.title(f"Experiment {run_id} - Learning Curves")
    plt.xlabel("Epoch")
    plt.ylabel("MSE (kg²)")
    plt.grid(alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()


# Baseline + All 30 Experiments
all_run_ids = [
    "BASE_NN_EARLY_STOPPING"
] + [
    run_id for run_id, topic, changes in all_experiments
]

print("Total experiments:", len(all_run_ids))

for run_id in all_run_ids:
    print(f"\nLearning Curve: {run_id}")
    plot_experiment(run_id)
```

```python

# Final Test Evaluation - Save and Reload Everything

import json
import numpy as np
import torch
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

BEST_MODEL_ID = "B10"

test_result_path = PROJECT_DIR / "FINAL_TEST_EVALUATION.json"
test_arrays_path = PROJECT_DIR / "FINAL_TEST_PREDICTIONS.npz"

# CASE 1: Previously saved predictions are available
if test_arrays_path.exists():

    saved = np.load(test_arrays_path)

    y_test = saved["y_test"]
    test_predictions = saved["test_predictions"]

    with open(test_result_path, "r") as f:
        test_result = json.load(f)

    print("Previously saved test results loaded.")

# CASE 2: First evaluation or predictions were not saved
else:

    print("Preparing final B10 test evaluation...")

    checkpoint = torch.load(
        MODEL_DIR / "B10.pt",
        map_location=device,
        weights_only=False
    )

    X_test = test_df[features]
    y_test = test_df[target].to_numpy(dtype=np.float64)

    X_test_processed = preprocessor.transform(X_test)

    X_test_tensor = torch.tensor(
        X_test_processed,
        dtype=torch.float32
    ).to(device)

    final_model = build_mlp(
        X_train_processed.shape[1],
        checkpoint["config"]["hidden"]
    ).to(device)

    final_model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    final_model.eval()

    with torch.no_grad():
        predictions_scaled = (
            final_model(X_test_tensor)
            .cpu()
            .numpy()
            .ravel()
        )

    test_predictions = (
        predictions_scaled * checkpoint["y_std"]
        + checkpoint["y_mean"]
    )

    test_mse = mean_squared_error(
        y_test,
        test_predictions
    )

    test_result = {
        "model": BEST_MODEL_ID,
        "validation_mse": float(checkpoint["best_val_mse"]),
        "test_mse": float(test_mse),
        "test_samples": len(y_test)
    }

    # Save evaluation results
    with open(test_result_path, "w") as f:
        json.dump(test_result, f, indent=4)

    # Save actual values and predictions
    np.savez(
        test_arrays_path,
        y_test=y_test,
        test_predictions=test_predictions
    )

    print("First test evaluation completed and saved.")

# Prepare residuals for visualization
residuals = y_test - test_predictions

print("\nFinal Test Results")
print("--------------------------")
print("Model:", BEST_MODEL_ID)
print("Validation MSE:",
      round(test_result["validation_mse"], 2), "kg²")
print("Test MSE:",
      round(test_result["test_mse"], 2), "kg²")
print("Test MAE:",
      round(np.mean(np.abs(residuals)), 2), "kg")
print("Test Samples:", len(y_test))
print("Predictions saved successfully!")
```
